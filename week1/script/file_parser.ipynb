```python
# Open the file
file = open("data.txt", "r")

# Read the content
content = file.read()

# Close the file
file.close()

# Count lines
lines = content.splitlines()

# Count words
words = content.split()

# Count characters
characters = len(content)

# Display results
print("===== File Parser =====")

print("Total Lines:", len(lines))

print("Total Words:", len(words))

print("Total Characters:", characters)
```
