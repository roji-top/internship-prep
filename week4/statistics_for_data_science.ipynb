## Statistics For Data Science


**Statistics** is like a toolkit we use to understand and make sense of information. It helps us collect, organize, analyze, and interpret data to find patterns, trends, and relationships in the world around us.

**Statistics** is the branch of **mathematics** that deals with collecting, analyzing, interpreting, presenting, and organizing data. It involves the study of methods for gathering, summarizing, and interpreting data to make informed decisions and draw meaningful conclusions.



### Why for Data Science

Statistics is the backbone of data science because it provides the mathematical and analytical tools required to extract insights from data.

- **Data Understanding and Summarization:**
  - Statistics helps in understanding data distributions, central tendencies (mean, median, mode), and variability (variance, standard deviation). 
  - This allows data scientists to summarize and preprocess data effectively before modeling

- **Data Sampling and Experimental Design:**
  - Since working with entire populations is often impractical, statistical sampling methods (random, stratified, etc.) help in selecting representative subsets.  
  - Experimental design ensures that data collection is unbiased and follows scientific principles.

- **Probability and Uncertainty Estimation:**
  - Many data science models, such as Bayesian inference and probabilistic machine learning models, rely on probability theory

- **Hypothesis Testing:**
    - Statistical tests (t-tests, chi-square tests, ANOVA) validate hypotheses about datasets, helping in data-driven decision-making




### Types of Statistics


## 1. Descriptive Statistics  
Descriptive statistics summarize and describe the main characteristics of a dataset. It does not make predictions or infer conclusions about a larger population.  

### Key Components of Descriptive Statistics  

#### 🔹 Measures of Central Tendency (Summarize the center of data)  
- **Mean**: The average of all values.  
- **Median**: The middle value when sorted.  
- **Mode**: The most frequently occurring value.  

#### 🔹 Measures of Dispersion (Describe the spread of data)  
- **Range**: Difference between maximum and minimum values.  
- **Variance**: Average squared difference from the mean.  
- **Standard Deviation**: Square root of variance, measures spread.  
- **Interquartile Range (IQR)**: Spread between Q1 (25th percentile) and Q3 (75th percentile).  

#### 🔹 Data Distribution & Shape  
- **Symmetric (Normal) Distribution**: Bell-shaped curve with mean = median = mode.  
- **Skewed Distribution**: Data is asymmetrical (right or left skewed).  
- **Kurtosis**: Measures how peaked or flat a distribution is.  

---  

## 2. Inferential Statistics  
Inferential statistics allows us to make predictions or draw conclusions about a population based on a sample.  

### Key Components of Inferential Statistics  

#### 🔹 Sampling Methods  
- **Random Sampling**: Every individual has an equal chance of selection.  
- **Stratified Sampling**: Population divided into groups, and samples are taken proportionally.  
- **Systematic Sampling**: Selecting every nth individual from a list.  
- **Cluster Sampling**: Population divided into clusters, then randomly selecting entire clusters.  

#### 🔹 Estimation & Confidence Intervals  
- **Point Estimation**: A single value estimate of a population parameter (e.g., sample mean estimating population mean).  
- **Confidence Intervals (CI)**: A range within which the true population parameter lies with a certain confidence level (e.g., 95% CI).  

#### 🔹 Hypothesis Testing  
- **Null Hypothesis (H₀)**: Assumes no effect or no difference.  
- **Alternative Hypothesis (H₁)**: Assumes an effect or difference exists.  
- **p-value**: Probability of obtaining observed results if H₀ is true. If p < 0.05, reject H₀.  

#### 🔹 Common Statistical Tests  
- **t-Test**: Compares means of two groups (independent or paired).  
- **Chi-Square Test**: Tests categorical variable independence.  
- **ANOVA (Analysis of Variance)**: Compares means of three or more groups.  
- **Regression Analysis**: Determines relationships between dependent and independent variables.

# Measurement Scales of Variables

Statistical variables can be measured at different levels, determining how data can be analyzed and interpreted. The four types of measurement scales are Nominal, Ordinal, Interval, and Ratio.


| Measurement Scale | Type of Variable | Description                                                                 | Examples                                      | Allowed Operations                                                                 |
|------------------|------------------|-----------------------------------------------------------------------------|-----------------------------------------------|-----------------------------------------------------------------------------------|
| **Nominal**      | Categorical      | Labels or names without any meaningful order.                               | Colors (Red, Blue), Nationality (USA, Canada) | Counting, Mode                                                                    |
| **Ordinal**      | Categorical      | Ordered categories, but differences between them are not neaningful.      | Education Level (High School < Bachelor's < Master's) | Counting, Mode, Median, Ranking                                                  |
| **Interval**     | Numerical        | Ordered with equal intervals, but no true zero point.                       | Temperature (°C), IQ Scores                   | Mean, Standard Deviation, Addition & Subtraction                                |
| **Ratio**        | Numerical        | Ordered with equal intervals and a true zero (absolute zero exists).         | Height, Weight, Income, Age                  | All arithmetic operations (Mean, Ratio, Multiplication, Division)                |

### 1. Explanation of Each Scale

#### **Nominal Scale (Categorical, No Order)**
Represents labels or names without any meaningful order.

Example:

- Gender: {Male, Female, Other}
- Blood Type: {A, B, AB, O}

#### **Ordinal Scale (Categorical, Ordered)**
Categories have a meaningful rank order, but the differences between them are not meaningful.

Example:

- Education Levels: {High School < Bachelor's < Master's < PhD}
- Difference between Bachelor's and Master's is not necessarily the same meaning as between High School and Bachelor's
- Customer Satisfaction: {Very Unsatisfied < Unsatisfied < Neutral < Satisfied < Very Satisfied}

#### **Interval Scale (Numerical, No True Zero)**
Ordered data with equal intervals, but no absolute zero (zero is arbitrary and does not mean "nothing").

Example:

- Temperature in Celsius/Fahrenheit (0°C does not mean "no temperature").
- 40°C is not "twice as hot" as 20°C - we cannot determine ratio
- IQ Scores (A person with an IQ of 140 is not "twice as intelligent" as someone with an IQ of 70).

#### **Ratio Scale (Numerical, True Zero)**

Ordered data with equal intervals and a meaningful zero (zero means "absence" of the quantity).

Example:

- Weight (0 kg means "no weight")
- Distance (0 meters means "no distance")
- Income ($0 means "no money")

**Allowed Operations:**

All arithmetic operations: Addition, Subtraction, Multiplication, Division, and Measures like Mean, Standard Deviation, Ratios

# Measure of Shape

## Skewness  Resources (https://www.geeksforgeeks.org/program-find-skewness-statistical-data/)

Skewness is a measure of the asymmetry of the probability distribution of a real-valued random variable about its mean. In simpler terms, it indicates whether the data is concentrated more on one side of the mean compared to the other side.

### 1. Symmetric Distribution (Skewness ≈ 0)
Appearance: The distribution looks the same on both sides of the center

Mean ≈ Median ≈ Mode

- Example: Heights of adult humans, measurement errors

### 2. Right (Positive) Skew
Appearance: The "tail" extends to the right

Mean > Median > Mode

Why called "right" skew: The tail points to the right on the number line

Real-world examples:
- Income distribution (few extremely high incomes)
- House prices
- Hospital stay durations

### 3. Left (Negative) Skew
Appearance: The "tail" extends to the left

Mean < Median < Mode

Why called "left" skew: The tail points to the left on the number line

Real-world examples:

- Age at retirement (most retire around same age, few very early)
- Exam scores when there's a minimum passing score
- Age at death in developed countries

```python
# Simulated income data (right-skewed)
incomes = np.random.lognormal(mean=3, sigma=0.7, size=1000)

plt.hist(incomes, bins=30, edgecolor='black')
plt.title('Income Distribution (Right Skewed)')
plt.axvline(np.mean(incomes), color='red', linestyle='--', label=f'Mean (${np.mean(incomes):.2f}k)')
plt.axvline(np.median(incomes), color='green', linestyle='--', label=f'Median (${np.median(incomes):.2f}k)')
plt.xlabel('Annual Income ($1000)')
plt.legend()
plt.show()

print(f"Mean income: ${np.mean(incomes):.2f}k")
print(f"Median income: ${np.median(incomes):.2f}k")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import skew

# Create three datasets with different skewness
normal_data = np.random.normal(0, 1, 1000)  # Symmetric
right_skewed = np.random.exponential(scale=1.0, size=1000)  # Right skewed
left_skewed = -np.random.exponential(scale=1.0, size=1000)  # Left skewed

# Plot them
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Normal distribution
axes[0].hist(normal_data, bins=30, edgecolor='black')
axes[0].set_title(f'Symmetric (Skew = {skew(normal_data):.2f})')
axes[0].axvline(np.mean(normal_data), color='red', linestyle='dashed', linewidth=1, label='Mean')
axes[0].axvline(np.median(normal_data), color='green', linestyle='dashed', linewidth=1, label='Median')

# Right skewed
axes[1].hist(right_skewed, bins=30, edgecolor='black')
axes[1].set_title(f'Right Skewed (Skew = {skew(right_skewed):.2f})')
axes[1].axvline(np.mean(right_skewed), color='red', linestyle='dashed', linewidth=1)
axes[1].axvline(np.median(right_skewed), color='green', linestyle='dashed', linewidth=1)

# Left skewed
axes[2].hist(left_skewed, bins=30, edgecolor='black')
axes[2].set_title(f'Left Skewed (Skew = {skew(left_skewed):.2f})')
axes[2].axvline(np.mean(left_skewed), color='red', linestyle='dashed', linewidth=1)
axes[2].axvline(np.median(left_skewed), color='green', linestyle='dashed', linewidth=1)

plt.legend()
plt.show()
```

# 1. What is Dispersion?

**Dispersion** refers to how spread out the data points are in a dataset. It helps us understand:

- Variability: How much individual data points differ from each other.
- Consistency: Whether data points are clustered closely or widely scattered.

## Why is Dispersion Important?
- Example: Two classes have the same average test score of 70.

     Class A: Scores = [68, 70, 72] → Low dispersion (consistent performance).

     Class B: Scores = [50, 70, 90] → High dispersion (some students did very well, some poorly).

- Conclusion: The average alone doesn’t tell the full story. Dispersion measures help quantify spread.



# 2. Measures of Dispersion

## Range
Definition: Difference between the maximum and minimum values.

Formula:

    Range = Max − Min

Example:

    Data = [5, 10, 15, 20, 25]
    
    Range = 25 - 5 = 20

**Limitations of Range**
- Only considers two extreme values (ignores the rest).
- Sensitive to outliers.

## Variance (σ² or s²)
Definition: Measures how far each data point is from the mean.




**Interpretation of Variance**

- Higher variance → Data points are more spread out.
- Lower variance → Data points are closer to the mean.

**Example Calculation**

Dataset = [2, 4, 6, 8]

Mean (μ) = (2+4+6+8)/4 = 5

Squared deviations: (2−5)²=9, 1, 1, 9
Variance (σ²) = (9 + 1 + 1 + 9)/4 = 5



#### Python Example (Variance)


```python
import numpy as np

data = [2, 4, 6, 8]

# Population variance (ddof=0)
var_pop = np.var(data, ddof=0)  
print("Population Variance:", var_pop)  # Output: 5.0

# Sample variance (ddof=1)
var_sample = np.var(data, ddof=1)  
print("Sample Variance:", var_sample)  
```

## Standard Deviation (σ or s)

Definition: Square root of variance (to bring units back to original scale).



```python
import numpy as np

data = [2, 4, 6, 8]

# Population SD (ddof=0)
std_pop = np.std(data, ddof=0)  
print("Population SD:", std_pop)  # Output: ~2.236

# Sample SD (ddof=1)
std_sample = np.std(data, ddof=1)  
print("Sample SD:", std_sample)  # Output: ~2.58
```

### Practical Applications: Case Study - Comparing Two Stocks

| Stock           | Mean Daily Return | Standard Deviation | Interpretation                          |
|-----------------|-------------------|--------------------|-----------------------------------------|
| **Stock A**     | 1.5%              | 0.5%               | Low SD → More stable, less risky        |
| **Stock B**     | 1.5%              | 2.0%               | High SD → More volatile, higher risk    |

### Key Insights:
1. **Same Mean, Different Risk**:  
   Both stocks have identical average returns (1.5%), but their risk profiles differ dramatically.

2. **Stock A (SD = 0.5%)**:  
   - Returns cluster tightly around 1.5%.  
   - 68% of daily returns fall between **1.0% and 2.0%** (Mean ± 1 SD).  
   - Ideal for conservative investors.

3. **Stock B (SD = 2.0%)**:  
   - Returns are widely dispersed.  
   - 68% of daily returns fall between **-0.5% and 3.5%**.  
   - Potential for higher gains but also larger losses.

```python
import numpy as np
import matplotlib.pyplot as plt

# Simulate stock returns
np.random.seed(42)
stock_A = np.random.normal(1.5, 0.5, 100)  # Low SD
stock_B = np.random.normal(1.5, 2.0, 100)  # High SD

plt.figure(figsize=(10, 5))
plt.plot(stock_A, label='Stock A (SD=0.5)', alpha=0.7)
plt.plot(stock_B, label='Stock B (SD=2.0)', alpha=0.7)
plt.axhline(1.5, color='red', linestyle='--', label='Mean (1.5%)')
plt.title("Stock Volatility Comparison")
plt.ylabel("Daily Return (%)")
plt.legend()
plt.show()
```

### Gaussian (Normal) Distribution
The Gaussian distribution, also called the Normal distribution, is a continuous probability distribution symmetric about its mean.

Key Properties:
- Symmetrical (bell-shaped curve)

- Mean = Median = Mode

- Defined by two parameters: Mean (μ) → Center of the distribution, Variance (σ²) → Spread of the distribution

- 68-95-99.7 Rule (Empirical Rule):

~68% of data falls within μ ± σ

~95% within μ ± 2σ

~99.7% within μ ± 3σ

## Python Example: Adult Heights (Gaussian (Normal) Distribution)
Assume adult heights follow a normal distribution with μ = 170 cm, σ = 10 cm.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parameters
mu, sigma = 170, 10

# Generate data
x = np.linspace(140, 200, 1000)
pdf = norm.pdf(x, mu, sigma)

# Plot
plt.figure(figsize=(10, 5))
plt.plot(x, pdf, 'b-', label=f'N(μ={mu}, σ²={sigma**2})')
plt.fill_between(x, pdf, where=(x >= 160) & (x <= 180), color='skyblue', alpha=0.5, label='P(160 ≤ X ≤ 180)')
plt.title("Gaussian Distribution (Heights in cm)")
plt.xlabel("Height (cm)")
plt.ylabel("Probability Density")
plt.legend()
plt.grid()
plt.show()

# Probability of height between 160-180 cm
prob = norm.cdf(180, mu, sigma) - norm.cdf(160, mu, sigma)
print(f"Probability (160 ≤ X ≤ 180): {prob:.2%}")
```
